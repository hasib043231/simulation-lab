```python
import numpy as np
import matplotlib.pyplot as plt

def monte_carlo_ellipse(N):

    x = np.random.uniform(0, 2, N)
    y = np.random.uniform(0, 1, N)


    inside_ellipse = (x**2 / 4) + y**2 <= 1


    rect_area = 2 * 1
    estimated_area = (np.sum(inside_ellipse) / N) * rect_area

    return x, y, inside_ellipse, estimated_area


N_values = [100, 1000, 10000]
true_area = np.pi / 2

print(f"{'N':>10} | {'Estimated Area':>15} | {'Error':>10}")
print("-" * 42)

for N in N_values:
    x, y, inside, est_area = monte_carlo_ellipse(N)
    error = abs(est_area - true_area)
    print(f"{N:10d} | {est_area:15.4f} | {error:10.4f}")


x, y, inside, est_area = monte_carlo_ellipse(1000)
plt.figure(figsize=(8, 4))
plt.scatter(x[inside], y[inside], color='blue', s=1, label='Inside')
plt.scatter(x[~inside], y[~inside], color='red', s=1, label='Outside')
plt.title(f"Monte Carlo Simulation (N=1000)\nEstimated Area: {est_area:.4f}")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
