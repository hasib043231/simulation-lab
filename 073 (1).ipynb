```python


import numpy as np
import matplotlib.pyplot as plt


np.random.seed(42)




regular_wait = np.random.uniform(2, 12, 3000)


busy_wait = np.random.normal(loc=8, scale=2, size=3000)


busy_wait = np.where(busy_wait < 0, 0, busy_wait)



plt.figure(figsize=(12,5))


plt.subplot(1,2,1)
plt.hist(regular_wait, bins=30, alpha=0.7)
plt.title("Regular Day Wait Times (Uniform)")
plt.xlabel("Minutes")
plt.ylabel("Frequency")


plt.subplot(1,2,2)
plt.hist(busy_wait, bins=30, alpha=0.7)
plt.title("Busy Day Wait Times (Normal)")
plt.xlabel("Minutes")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()



def print_stats(data, name):
    print(f"\n{name} Statistics:")
    print(f"Mean: {np.mean(data):.2f}")
    print(f"Median: {np.median(data):.2f}")
    print(f"Standard Deviation: {np.std(data):.2f}")
    print(f"Maximum Wait Time: {np.max(data):.2f}")

print_stats(regular_wait, "Regular Day")
print_stats(busy_wait, "Busy Day")
```

```python
import numpy as np


d1 = 3
d2 = 7

A = np.array([
    [d1, d2, 1],
    [2, d1 + d2, 3],
    [4, 5, d2]
])

print("Matrix A:\n", A)


print("\n--- Matrix A Analysis ---")
print("Shape:", A.shape)

det_A = np.linalg.det(A)
print("Determinant:", det_A)

rank_A = np.linalg.matrix_rank(A)
print("Rank:", rank_A)

eig_A = np.linalg.eigvals(A)
print("Eigenvalues:", eig_A)

if det_A != 0:
    inv_A = np.linalg.inv(A)
    print("Inverse of A:\n", inv_A)
else:
    print("Matrix A is not invertible")


B = A.copy()
B[0,0] += 1
print("\nMatrix B (changed A[0,0] + 1):\n", B)



print("\n--- Matrix B Analysis ---")
print("Shape:", B.shape)

det_B = np.linalg.det(B)
print("Determinant:", det_B)

rank_B = np.linalg.matrix_rank(B)
print("Rank:", rank_B)

eig_B = np.linalg.eigvals(B)
print("Eigenvalues:", eig_B)

if det_B != 0:
    inv_B = np.linalg.inv(B)
    print("Inverse of B:\n", inv_B)
else:
    print("Matrix B is not invertible")
```
