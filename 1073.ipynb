```python
import numpy as np
from scipy import stats


A, B, C, D = 1, 0, 7, 3
X, Y = A + B, C + D

data_a = np.array([48+X, 52-Y, 49+A, 51-B, 50+C, 47+D, 53-A, 50+B])
t_stat_a, p_val_a = stats.ttest_1samp(data_a, 50)


mach_a = np.array([45+X, 47+A, 46+B, 48+C, 44+D, 49+X])
mach_b = np.array([43+Y, 42+A, 44+B, 41+C, 45+D, 43+Y])

t_equal, p_equal = stats.ttest_ind(mach_a, mach_b, equal_var=True)
t_unequal, p_unequal = stats.ttest_ind(mach_a, mach_b, equal_var=False)

print(f"Part A: t={t_stat_a:.3f}, p={p_val_a:.3f}")
print(f"Part B (Equal Var): t={t_equal:.3f}, p={p_equal:.3f}")
print(f"Part B (Unequal Var): t={t_unequal:.3f}, p={p_unequal:.3f}")
```
