```python
import random
import math
import numpy as np

def run_pi_simulation(student_id, full_name, multiplier=1):


    R = int(str(student_id)[-2:])
    if R == 0:
        R = 10


    radius = R / 10.0


    name_length = len(full_name.replace(" ", ""))


    base_points = 200 * name_length
    total_points = base_points * multiplier


    half_points = total_points // 2
    remainder = total_points - half_points

    points_inside = 0


    for _ in range(half_points):
        x = random.uniform(0, radius)
        y = random.uniform(0, radius)

        if x**2 + y**2 <= radius**2:
            points_inside += 1

    x_np = np.random.rand(remainder) * radius
    y_np = np.random.rand(remainder) * radius

    inside_np = np.sum((x_np**2 + y_np**2) <= radius**2)

    points_inside += inside_np


    estimated_pi = 4 * (points_inside / total_points)


    error = abs(math.pi - estimated_pi)

    return f"[0, {radius:.1f}]", total_points, points_inside, estimated_pi, error



STUDENT_ID = "73"
FULL_NAME = "Hasibur Rahman"

print(f"--- Monte Carlo Simulation for {FULL_NAME} (ID: {STUDENT_ID}) ---")
print(f"{'Range Used':<12} | {'Total Points':<12} | {'Points Inside':<13} | {'Estimated Pi':<13} | {'Error from math.pi'}")
print("-" * 80)


for i in range(1, 6):
    rng, total, inside, est_pi, err = run_pi_simulation(STUDENT_ID, FULL_NAME, multiplier=i)

    print(f"{rng:<12} | {total:<12} | {inside:<13} | {est_pi:<13.6f} | {err:.6f}")
```
