```python

import numpy as np


np.random.seed(0)
A = np.random.randint(1, 10, (3, 3))
B = np.random.randint(1, 10, (3, 3))

print("Matrix A:\n", A)
print("\nMatrix B:\n", B)


addition = A + B
subtraction = A - B
multiplication = A @ B

print("\nA + B:\n", addition)
print("\nA - B:\n", subtraction)
print("\nA × B:\n", multiplication)


det_A = np.linalg.det(A)
rank_A = np.linalg.matrix_rank(A)

print("\nDeterminant of A:", det_A)
print("Rank of A:", rank_A)


if det_A != 0:
    inv_A = np.linalg.inv(A)
    print("\nInverse of A:\n", inv_A)
else:
    print("\nMatrix A is singular, inverse not possible.")
```

```python


import numpy as np
import matplotlib.pyplot as plt


np.random.seed(1)
vec1 = np.random.rand(15)
vec2 = np.random.rand(15)


plt.figure(figsize=(8,5))
plt.plot(vec1, marker='o', label='Vector 1')
plt.plot(vec2, marker='s', label='Vector 2')


plt.xlabel("Index")
plt.ylabel("Value")
plt.title("Comparison of Two Random Vectors")
plt.legend()
plt.grid(True)

plt.show()
```
