```python
import numpy as np


d1 = 3
d2 = 7

A = np.array([
    [d1, d2, 1],
    [2, d1 + d2, 3],
    [4, 5, d2]
])

print("Matrix A:\n", A)


print("\n--- Matrix A Analysis ---")
print("Shape:", A.shape)

det_A = np.linalg.det(A)
print("Determinant:", det_A)

rank_A = np.linalg.matrix_rank(A)
print("Rank:", rank_A)

eig_A = np.linalg.eigvals(A)
print("Eigenvalues:", eig_A)

if det_A != 0:
    inv_A = np.linalg.inv(A)
    print("Inverse of A:\n", inv_A)
else:
    print("Matrix A is not invertible")


B = A.copy()
B[0,0] += 1
print("\nMatrix B (changed A[0,0] + 1):\n", B)



print("\n--- Matrix B Analysis ---")
print("Shape:", B.shape)

det_B = np.linalg.det(B)
print("Determinant:", det_B)

rank_B = np.linalg.matrix_rank(B)
print("Rank:", rank_B)

eig_B = np.linalg.eigvals(B)
print("Eigenvalues:", eig_B)

if det_B != 0:
    inv_B = np.linalg.inv(B)
    print("Inverse of B:\n", inv_B)
else:
    print("Matrix B is not invertible")
```
